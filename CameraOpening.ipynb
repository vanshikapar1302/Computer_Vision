```python
import cv2
```

```python
cap = cv2.VideoCapture(0)
while True:
    flag, frame = cap.read()
    if not flag:
        print("Camera not accessible")
        break

    cv2.imshow("My Camera", frame)
    key = cv2.waitKey(25) & 0xFF

    if key == 27: 
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import yt_dlp
```

```python
URL = "https://www.youtube.com/watch?v=0M4iSY33lPw"
ydl_opts = {
    'format': 'best[ext=mp4]'
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    info = ydl.extract_info(URL, download=False)
    video_url = info['url']
    cap = cv2.VideoCapture(video_url)
    cap = cv2.VideoCapture(0)
    while True:
        flag, frame = cap.read()
        if not flag:
            print("Camera not accessible")
            break
        cv2.imshow("My Camera", frame)
        key = cv2.waitKey(25) & 0xFF
        if key == 27: 
            break

cap.release()
cv2.destroyAllWindows()
 
```
