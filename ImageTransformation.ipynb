```python
# CamScanner
# Image Transformation-Translate,Rotate,Resize
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv2.imread("z1.png")
```

```python
img.shape
```

```python
plt.imshow(img)
```

```python
rows,cols=img.shape[:2]
```

```python
#Translate image
# create translation matrix
#Format:[[1,0,tx],[0,1,ty]]
#tx - shift along x-axis
#ty - shift along y-axis
matrix=np.float32([[1,0,-100],[0,1,10]])

#Apply translation
translated=cv2.warpAffine(img,matrix,(cols,rows))
plt.imshow(translated)
plt.show()

```

```python
#Rotation
#Centre of rotation
centre=(rows//2,cols//2)
#params:(centre,angle,scale)
# angle - rotation in degrees(positive=counter clockwise)
#scale-zoom(default=1)
matrix=cv2.getRotationMatrix2D(centre,-6,1)
rotated = cv2.warpAffine(img,matrix,(cols,rows))
plt.imshow(rotated)
plt.show()
```

```python
# Resize image
resized=cv2.resize(rotated,(1000,850))
plt.imshow(resized)
plt.show()
```

```python
# rotated[start_y:end_y,start_x:end_x]
# first=rows(height-y)
# second=cols(width-x)
cropped=rotated[200:1400,950:1900]
plt.imshow(cropped)
plt.show()
```
